```python
def Lagrange(x,y,xx):
    t_sum = 0
    n = len(x)
    for i in range(n):
        product = y[i]
        for j in range(n):
            if i != j :
                product = product*(xx-x[j])/(x[i]-x[j])
        t_sum = t_sum+product 
    return t_sum
```

```python
Lagrange([1,2,3],[1,4,9],0.5)
```
